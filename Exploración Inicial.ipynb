```python
avg_len = 0
avg_word_count = 0
for i in range(len(sylo)):
    length = len(sylo[i])
    word_count = sylo[i].split()
    avg_len += length
    avg_word_count += len(word_count)

print("Cantidad de instancias: {}".format(len(sylo)))
print("Longitud promedio (caracteres): {}".format(round(avg_len/len(sylo), 3)))
print("Longitud promedio (palabras): {}".format(round(avg_word_count/len(sylo), 3)))
```

```python
log_words = ['anything', 'no', 'every', 'nothing', 'all', 'exists', 'therefore', 'consequently', 'everything', 'there exists']

norm_sylo = [_.lower() for _ in sylo]
norm_sylo = [_.replace(',', '') for _ in norm_sylo]
norm_sylo = [_.replace('.', '') for _ in norm_sylo]

log_value = 0
for _ in norm_sylo:
    current = _.split()
    aux = 0
    for word in current:
        if word in log_words:
            aux += 1
    if aux != 0:
        log_value+=1

print("Palabras lógicamente relevantes dentro del DS: {}".format(log_words))
print("Cantidad de instancias que tienen un conector: {}".format(log_value))
```
